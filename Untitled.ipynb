## Dataset Discovery and EDA Notebook

```python
# Imports

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
chart_events = pd.read_csv("Data/MIMIC-IV-demo/icu/chartevents.csv.gz")
icu_stays = pd.read_csv("Data/MIMIC-IV-demo/icu/icustays.csv.gz")

print("Chartevents shape:", chart_events.shape)
print("ICU stays shape:", icu_stays.shape)
print("\nChartevents columns:", chart_events.columns.tolist())
```

```python
ITEM_IDS = {
    220045: "heart_rate",
    220277: "spo2",
    # Invasive BP
    220050: "systolic_bp_inv",
    220051: "diastolic_bp_inv",
    220052: "mean_bp_inv",
    # Non-invasive BP
    220179: "systolic_bp_cuff",
    220180: "diastolic_bp_cuff",
    220181: "mean_bp_cuff"
}

# Keep only rows we care about
df = chart_events[chart_events["itemid"].isin(ITEM_IDS.keys())].copy()
df["measurement"] = df["itemid"].map(ITEM_IDS)

print("Rows after filter:", df.shape[0])
print(df["measurement"].value_counts())
```

```python
df = df.dropna(subset=["valuenum"])
df = df[df["warning"] != 1]

# Physiologically valid ranges
valid_ranges = {
    "heart_rate":        (20, 300),
    "spo2":              (50, 100),
    "systolic_bp_inv":   (50, 250),
    "diastolic_bp_inv":  (20, 150),
    "mean_bp_inv":       (30, 200),
    "systolic_bp_cuff":  (50, 250),
    "diastolic_bp_cuff": (20, 150),
    "mean_bp_cuff":      (30, 200)
}

mask = pd.Series(True, index=df.index)
for measurement, (low, high) in valid_ranges.items():
    bad = (df["measurement"] == measurement) & ~df["valuenum"].between(low, high)
    mask = mask & ~bad

df = df[mask]
print("Rows after range filter:", df.shape[0])
```

```python
df = df[["subject_id", "stay_id", "charttime", "measurement", "valuenum"]]

df_pivot = df.pivot_table(
    index=["subject_id", "stay_id", "charttime"],
    columns="measurement",
    values="valuenum",
    aggfunc="mean"
).reset_index()

df_pivot.columns.name = None
df_pivot["charttime"] = pd.to_datetime(df_pivot["charttime"])

# Combine invasive + cuff into single columns
# Prefer invasive when both exist, fall back to cuff
df_pivot["systolic_bp"]  = df_pivot["systolic_bp_inv"].combine_first(df_pivot["systolic_bp_cuff"])
df_pivot["diastolic_bp"] = df_pivot["diastolic_bp_inv"].combine_first(df_pivot["diastolic_bp_cuff"])
df_pivot["mean_bp"]      = df_pivot["mean_bp_inv"].combine_first(df_pivot["mean_bp_cuff"])

# Drop the original split columns
df_pivot = df_pivot.drop(columns=[
    "systolic_bp_inv", "systolic_bp_cuff",
    "diastolic_bp_inv", "diastolic_bp_cuff",
    "mean_bp_inv", "mean_bp_cuff"
])

print("Shape after pivot:", df_pivot.shape)
print(df_pivot.head())
```

```python
df_pivot = df_pivot.dropna(
    subset=["heart_rate", "spo2", "systolic_bp", "diastolic_bp", "mean_bp"]
)
print("Clean rows:", df_pivot.shape[0])
```

```python
df_pivot = df_pivot.sort_values(["subject_id", "stay_id", "charttime"])

grp = df_pivot.groupby(["subject_id", "stay_id"])["heart_rate"]

# Rolling window features (window=5 measurements)
df_pivot["hr_rolling_mean"] = grp.transform(lambda x: x.rolling(5, min_periods=2).mean())
df_pivot["hr_rolling_std"]  = grp.transform(lambda x: x.rolling(5, min_periods=2).std())
df_pivot["hr_rolling_min"]  = grp.transform(lambda x: x.rolling(5, min_periods=2).min())
df_pivot["hr_rolling_max"]  = grp.transform(lambda x: x.rolling(5, min_periods=2).max())

# Beat-to-beat difference (proxy for HRV)
df_pivot["hr_diff"]         = grp.transform(lambda x: x.diff())
df_pivot["hr_diff_abs"]     = df_pivot["hr_diff"].abs()

# Lag features
df_pivot["hr_lag1"]         = grp.transform(lambda x: x.shift(1))
df_pivot["hr_lag2"]         = grp.transform(lambda x: x.shift(2))

grp_spo2 = df_pivot.groupby(["subject_id", "stay_id"])["spo2"]

df_pivot["spo2_rolling_mean"] = grp_spo2.transform(lambda x: x.rolling(5, min_periods=2).mean())
df_pivot["spo2_rolling_std"]  = grp_spo2.transform(lambda x: x.rolling(5, min_periods=2).std())
df_pivot["spo2_diff"]         = grp_spo2.transform(lambda x: x.diff())

# Drop rows with NaN from rolling/lag
df_pivot = df_pivot.dropna()
print("Shape after feature engineering:", df_pivot.shape)
```

```python
df_final = df_pivot.merge(
    icu_stays[["subject_id", "stay_id", "first_careunit", "los"]],
    on=["subject_id", "stay_id"],
    how="left"
)

# Encode care unit as a categorical
df_final["careunit_code"] = pd.Categorical(df_final["first_careunit"]).codes

print(df_final.shape)
print(df_final["first_careunit"].value_counts())
```

```python
fig, axes = plt.subplots(3, 3, figsize=(15, 8))

cols = ["heart_rate", "systolic_bp", "diastolic_bp", 
        "hr_rolling_std", "hr_rolling_mean", "hr_rolling_min",
        "hr_rolling_max", "hr_diff_abs", "mean_bp"]

for ax, col in zip(axes.flatten(), cols):
    df_final[col].hist(bins=50, ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()

print(df_final.describe())
```
